# **VERİ DÖNÜŞTÜRME**

**Verileri Toplama**: İstatistiksel analiz veya benzer içerikleri gruplama amacı doğrultusunda bilgilerin toplanıp özetlendiği işlemlerdir. Tavsiye sistemlerinde ve hedef kitle çalışmalarında kullanılır.

```python
df['Total_Power'] = df['HP'] + df['Attack'] + df['Defense'] + df['Sp. Atk'] + df['Sp. Def'] + df['Speed']
df[['Name', 'Total_Power']].head(3)
```

**Yeni Nitelik Oluşturma (Attribute Construction)**: Var olan nitelikleri daha iyi temsil edecek nitelikler oluşturma işlemidir.

```python
df['Attack_Defense_Ratio'] = df['Attack'] / (df['Defense'] + 1) # Sıfıra bölünme hatasını önlemek için +1
df[['Name', 'Attack_Defense_Ratio']].head(3)
```

**Kesimlere Ayırma (Data Discretization)**: Sürekli verilerin (Yaş, Hız vb.) belirli aralıklı etiketlerle kategorik verilere dönüştürülmesi.

```python
bins = [0, 50, 90, 200]
labels = ['Low Speed', 'Medium Speed', 'Very Fast']
df['Speed_Category'] = pd.cut(df['Speed'], bins=bins, labels=labels)

df[['Name', 'Speed', 'Speed_Category']].head(3)
```

**Veri Normalleştirme**: Farklı kaynaklardan gelen farklı büyüklükteki değerlerin aynı aralıkta temsil edilmesini sağlar

```python
numeric_cols = ['HP', 'Attack', 'Defense', 'Speed']
```

**Min-Max Scaler**: Sayısal verileri belirli bir aralığa, genellikle [0, 1] aralığına doğrusal olarak dönüştürmek için kullanılan temel bir ön işleme tekniğidir.


```python
minmax_scaler = MinMaxScaler()
df_minmax = df.copy()
df_minmax[numeric_cols] = minmax_scaler.fit_transform(df[numeric_cols])
df_minmax[numeric_cols].head(3)
```

**Z-Score**: Bir veri noktasının veri kümesinin ortalamasından kaç standart sapma uzaklıkta olduğunu gösteren istatistiksel bir ölçüdür.


```python
z_scaler = StandardScaler()
df_zscore = df.copy()
df_zscore[numeric_cols] = z_scaler.fit_transform(df[numeric_cols])
df_zscore[numeric_cols].head()
```

> **Korelasyon**: Verilerin birbiriyle olan ilişkisi. Bilgi Keşfi (KDD) süreçlerinde korelasyon, iki veya daha fazla değişken (sütun/özellik) arasındaki doğrusal ilişkinin yönünü ve gücünü gösteren istatistiksel bir ölçüdür.
> Öznitelik seçiminde korelasyon olabildiğince düşük (0'a yakın) tutulur. 

```python
numeric_cols = ['HP', 'Attack', 'Sp. Atk', 'Defense', 'Sp. Def', 'Speed', 'Total_Power', 'Attack_Defense_Ratio', 'Legendary']

# Filtre Yöntemi: Pearson Korelasyon Matrisi Hesaplama
corr_matriks = df[numeric_cols].corr(numeric_only=True)
corr_matriks.corr()
```

**Filtreler Yöntemi (Filters)**: Korelasyon katsayısına (Pearson Matrix) bakılarak öznitelik seçilmesi mantığına dayanır. Tüm modeli çalıştırmadığı için süreç olarak hızlıdır. Ancak tekrarlayan özelliklerin seçilmesine yol açabilir.

```python
df_numeric = df[numeric_cols]

X = df_numeric.drop(columns=['Legendary','Total_Power','Attack_Defense_Ratio']) # Bağımsız değişkenler
corr_matrix = X.corr().abs() #Pearson kolerasyon matrisi (mutlak değer-absolute)

# Simetrik tekrarları ve köşegendeki 1.0 değerlerini engellemek için Üst Üçgen Matrisini Alınır
upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

upper_tri
```

```python
# Belirlenen Eşik Değerinden (Threshold) Yüksek Olan Sütunları Bul
threshold = 0.5
to_drop = []

# Üst üçgen matrisindeki her bir sütun adını tek tek geziyoruz
for column in upper_tri.columns:
    # Sütundaki değerlerden herhangi biri belirlenen eşik değerinden (0.85) büyük mü?
    if (upper_tri[column] > threshold).any():
        # Eğer büyük bir değer varsa, bu sütun adını silinecekler listesine ekliyoruz
        to_drop.append(column)
```

```python
X_filtered = X.drop(columns=to_drop)
X_filtered = X_filtered.corr().abs()
X_filtered = X_filtered.where(np.triu(np.ones(X_filtered.shape), k=1).astype(bool))

X_filtered
```

**Paketleyiciler Yöntemi (Wrappers)**: Özelliklerin tek tek denenerek (ekleme, çıkarma veya özyinelemeli / Recursive) model üzerindeki etkisinin görüldüğü ve en iyi performans veren alt kümenin seçildiği yöntemdir. Süreç çok uzun sürer.   

```python
from sklearn.feature_selection import RFE # Recursive Feature Elimination
from sklearn.linear_model import LogisticRegression

# Bağımsız değişkenler (Öznitelikler) ve Hedef Değişken (Legendary: True/False)
X = df[['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']]
y = df['Legendary']

model = LogisticRegression(max_iter=1000)
rfe = RFE(estimator=model, n_features_to_select=3) # En iyi 3 özelliği seç
rfe.fit(X, y)

for i, col in enumerate(X.columns):
    print(f"{col}: Seçildi mi? -> {rfe.support_[i]} (Sıralama: {rfe.ranking_[i]})")
```

```python
from sklearn.ensemble import RandomForestClassifier

# Gömülü Yöntem: Random Forest kullanarak öznitelik önem skorlarını model içinde çıkarma
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X, y)

importances = pd.Series(rf_model.feature_importances_, index=X.columns)

importances.sort_values(ascending=False)
```
